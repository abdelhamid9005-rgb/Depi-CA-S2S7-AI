```python
# 1=>
def multiplication_table(n):
    """
    Print the multiplication table (1 through 10) of a given number.
    
    Returns None
        This function only prints the table; it does not return a value.
    """
    for i in range(1, 11):
        print(f"{n} * {i} = {n * i}")


multiplication_table(9)
```

```python
# 2=>
def is_prime(n):
    """
    Check whether a number is prime.
    
    Returns bool
        True if n is a prime number, False otherwise.
    """
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

print("Twin Prime:")
for num in range(2, 1000):
    if is_prime(num) and is_prime(num + 2):
        print(num, "and", num + 2)
```

```python
# 3=>
def primeFactors(n):
    """
    Find the prime factors of a given number.

    Returns list of int
        A list containing the prime factors of n, in ascending order,
        including repeated factors (e.g. primeFactors(56) -> [2, 2, 2, 7]).
    """
    factors = []
    divisor = 2
    while divisor * divisor <= n:
        while n % divisor == 0:
            factors.append(divisor)
            n //= divisor
        divisor += 1
    if n > 1:
        factors.append(n)
    return factors

primeFactors(56)

```

```python
# 4=>
def decToBin(n):
    """
    Convert a decimal (base-10) number to its binary representation.

    Returns str
        The binary representation of n as a string.
    """
    if n == 0:
        return "0"
    binary = ""
    while n > 0:
        binary = str(n % 2) + binary
        n //= 2
    return binary

print(decToBin(11))
```

```python
# 5=>
def is_perfect(n):
    """
    Check whether a number is a perfect number.
    A perfect number is a positive integer that equals the sum of its
    proper divisors (all divisors excluding the number itself).

    Returns bool
        True if n is a perfect number, False otherwise.
    """
    total = 0
    for i in range(1, n):
        if n % i == 0:
            total += i
    return total == n


def perfectNums(start, end):
    """
    Print all perfect numbers within a given inclusive range.
 
    Returns
    None
        This function only prints the perfect numbers; it does not return a value.
    """
    for n in range(start, end + 1):
        if is_perfect(n):
            print(n)


perfectNums(0, 100)
```
