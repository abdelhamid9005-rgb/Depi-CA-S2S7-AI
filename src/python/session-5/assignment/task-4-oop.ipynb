```python
# 1=>
class Task1:
    """
    Task1 class handles printing the multiplication table
    of a given number.

    """
    def __init__(self, number):
        """
        Initialize the Task1 object.
        Args:
            number (int): The number to generate the multiplication table for.
        """
        self.number = number

    def print_multiplication_table(self):
        """
        Print the multiplication table of the stored number, from 1 to 10.
        Returns: None
        """
        for i in range(1, 11):
            print(f"{self.number} * {i} = {self.number * i}")

task1 = Task1(9)
task1.print_multiplication_table()
```

```python
# 2=>
class Task2:
    """
    Task2 class handles finding and printing twin prime pairs
    below a given limit.
    Two consecutive odd numbers that are both prime are
    known as twin primes (e.g. 3 and 5, 11 and 13).

    """
    def __init__(self, limit):
        """
        Initialize the Task2 object.
        Args: limit (int): The upper bound (exclusive) for the search.
        """
        self.limit = limit

    def is_prime(self, n):
        """
        Determine whether a number is prime.
        Args: n (int): The number to check.
        Returns: bool: True if n is prime, False otherwise.
        """
        if n < 2:
            return False
        for i in range(2, int(n ** 0.5) + 1):
            if n % i == 0:
                return False
        return True

    def print_twin_primes(self):
        """
        Print all twin prime pairs below the stored limit.
        Returns: None
        """
        print("Twin Prime:")
        for num in range(2, self.limit):
            if self.is_prime(num) and self.is_prime(num + 2):
                print(num, "and", num + 2)

task2 = Task2(1000)
task2.print_twin_primes()
```

```python
# 3=>
class Task3:
    """
    Task3 class handles finding the prime factors of a given number.
    Attributes: number (int): The number to factorize.
    """

    def __init__(self, number):
        """
        Initialize the Task3 object.
        Args:  number (int): The number whose prime factors will be found.
        """
        self.number = number

    def get_prime_factors(self):
        """
        Find the prime factors of the stored number.
        Returns: list: A list of the prime factors, including repeated factors
                  (e.g. Task3(56).get_prime_factors() -> [2, 2, 2, 7]).
        """
        n = self.number
        factors = []
        divisor = 2
        while divisor * divisor <= n:
            while n % divisor == 0:
                factors.append(divisor)
                n //= divisor
            divisor += 1
        if n > 1:
            factors.append(n)
        return factors

task3 = Task3(56)
print(task3.get_prime_factors())
```

```python
# 4=>
class Task4:
    """
    Task4 class handles converting a decimal number
    to its binary representation.
    Attributes: number (int): The decimal number to convert.
    """

    def __init__(self, number):
        """
        Initialize the Task4 object.
        Args: number (int): A non-negative decimal integer to convert.
        """
        self.number = number

    def convert_to_binary(self):
        """
        Convert the stored decimal number to a binary string.
        Returns: str: The binary representation of the number
                 (e.g. Task4(11).convert_to_binary() -> "1011").
        """
        n = self.number
        if n == 0:
            return "0"
        binary = ""
        while n > 0:
            binary = str(n % 2) + binary
            n //= 2
        return binary

task4 = Task4(11)
print(task4.convert_to_binary())
```

```python
# 5=>
class Task5:
    """
    Task5 class handles finding all perfect numbers within a given range.
    A perfect number equals the sum of its proper divisors
    (all divisors excluding the number itself).

    Attributes:
        start (int): The start of the range (inclusive).
        end (int): The end of the range (inclusive).
    """

    def __init__(self, start, end):
        """
        Initialize the Task5 object.
        Args:
            start (int): The start of the range (inclusive).
            end (int): The end of the range (inclusive).
        """
        self.start = start
        self.end = end

    def is_perfect(self, n):
        """
        Determine whether a number is a perfect number.
        Args: n (int): The number to check.
        Returns: bool: True if n is a perfect number, False otherwise.
        """
        total = 0
        for i in range(1, n):
            if n % i == 0:
                total += i
        return total == n

    def print_perfect_numbers(self):
        """
        Print all perfect numbers within the stored range.
        Returns: None
        """
        for n in range(self.start, self.end + 1):
            if self.is_perfect(n):
                print(n)

task5 = Task5(0, 100)
task5.print_perfect_numbers()
```
